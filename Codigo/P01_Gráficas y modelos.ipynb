```python
x = [0, 1, 2, 3, 4]
y = [7, 4, 1, -2, -5]

# Definir la ecuación de primer grado y = mx + b
def lineal(z):
    return -3*z + 7

# Valores de x
z = np.linspace(-10, 10, 400)
salida = lineal(z)

plt.figure(figsize=(6,6))
plt.scatter(x, y, marker='o', label='Trazar puntos', color='red', linewidth=3)
plt.plot(z, salida, label='y = -3x + 7', color = 'blue', linestyle = '--', linewidth = 2)
plt.xlabel('x')
plt.ylabel('y')
plt.axhline(0, color='black',linewidth=1)
plt.axvline(0, color='black',linewidth=1)
plt.grid(color = 'gray', linestyle = '--', linewidth = 1)
plt.xlim(-5, 5)
plt.ylim(-8, 8)
plt.legend()
plt.show()
```

#### <font color='red'> 2. Intersecciones de una gráfica</font>

```python
px = [0, 2, -2]
py = [0, 0, 0]

def funcion(x):
    return x**3-4*x

# Valores de x
x = np.linspace(-10, 10, 400)
y = funcion(x)

plt.figure(figsize=(6,6))
plt.scatter(px, py, marker='o', label='Intersecciones', color='red', linewidth=3)
plt.plot(x, y, label='y = x^3-4x', color = 'green', linestyle = '--', linewidth = 2)
plt.xlabel('x')
plt.ylabel('y')
plt.axhline(0, color='black',linewidth=1)
plt.axvline(0, color='black',linewidth=1)
plt.grid(color = 'gray', linestyle = '--', linewidth = 1)
plt.xlim(-4, 4)
plt.ylim(-4, 4)
plt.legend()
plt.show()
```

#### Resolver una ecuación algebraica con una incógnita

```python
# Cargar librería con funciones específicas
from sympy import symbols, Eq, solve

# Definir la incógnita
x = symbols('x')

# Definir la ecuación
ecuacion = Eq(x**3-4*x, 0)

# Resolver la ecuación
solucion = solve(ecuacion, x)
print("Solución para la ecuación x^3-4x = 0: x =", solucion)
```

```python
# Cargar librería con funciones específicas
from sympy import symbols, Eq, solve

# Definir la incógnita
y = symbols('y')

# Definir la ecuación
ecuacion = Eq(y, 0)

# Resolver la ecuación
solucion = solve(ecuacion, x)
print("Solución para la ecuación y = 0: y =", solucion)
```

#### Resolver un sistema de ecuación con dos incógnitas

```python
# Definir ambas incógnitas
x, y = symbols('x y')

# Definir el sistema de ecuaciones
ecuacion1 = Eq(y, x**3 - 4*x)
ecuacion2 = Eq(y, 0)

# Resolver el sistema completo
soluciones = solve((ecuacion1, ecuacion2), (x, y))

print("Puntos de intersección encontrados:")
for i, sol in enumerate(soluciones, start=1):
    print(f"Punto {i}: (x = {sol[0]}, y = {sol[1]})")
```

#### <font color='red'> 4. Puntos de intersección</font>


```python
px = [2, -1]
py = [1, -2]

def funcion1(x):
    return x**2-3

def funcion2(x):
    return x-1

# Valores de x
x = np.linspace(-10, 10, 400)
y = funcion1(x)
z = funcion2(x)

plt.figure(figsize=(6,6))
plt.plot(x, y, label='y = x^2-3', color = 'blue', linestyle = '--', linewidth = 2)
plt.plot(x, z, label='y = x-1', color = 'black', linestyle = '--', linewidth = 2)
plt.scatter(px, py, marker='o', label='Puntos de intersección', color='red', linewidth=5)
plt.xlabel('x')
plt.ylabel('y')
plt.axhline(0, color='black',linewidth=1)
plt.axvline(0, color='black',linewidth=1)
plt.grid(color = 'gray', linestyle = '--', linewidth = 1)
plt.xlim(-3, 3)
plt.ylim(-3, 3)
plt.legend()
plt.show()
```

#### Resolver un sistema de ecuaciones de 2 incógnitas

```python
# Utilizamos la misma librería y funciones por lo que no hay que volver a cargar

# Definir las incógnitas
x, y = symbols('x y')

# Definir las ecuaciones
ecuacion1 = Eq(x**2 - y, 3)
ecuacion2 = Eq(x - y, 1)

# Resolver el sistema de ecuaciones
soluciones = solve((ecuacion1, ecuacion2), (x, y))

print("Soluciones para el sistema de ecuaciones:")
# Iteramos sobre la lista de puntos de intersección hallados
for i, sol in enumerate(soluciones, start=1):
    print(f"Punto {i}: x = {sol[0]}, y = {sol[1]}")
```
